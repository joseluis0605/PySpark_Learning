### Comparación código usando RDD y Dataframe

```python
# importamos librerias
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
# creamos sesión
spark = (SparkSession.builder
         .appName("aplicacion")
         .master("local[*]")
         .config("spark.driver.memory", "2g")
         .getOrCreate() 
)
```

### RDD

```python
# leemos el dataset
rdd = spark.sparkContext.textFile("04_usuarios.csv")

# obtenemos cabeceras
rrd_cabeceras = rdd.first()
print(rrd_cabeceras)

# eliminamos cabeceras
rdd_sin_cabeceras = rdd.filter(lambda x : x != rrd_cabeceras)

# troceamos csv
rdd_troceado = rdd_sin_cabeceras.map(lambda row : row.split(","))

# filtramos por mayores de 30 años
rdd_mayores30 = rdd_troceado.filter(lambda row : int(row[2])>30)

# hacemos recuento
rdd_mayores30.count()
```

### DATAFRAME

```python
# leemos csv
df = spark.read.csv("04_usuarios.csv", header = True, sep = ",")

# filtramos
df_filtrado = df.filter(F.col("edad") > 30)

# contamos
df_filtrado.count()
```
